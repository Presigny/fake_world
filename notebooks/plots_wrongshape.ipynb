# Weight analysis

```python
threshold = 1 # minimal number of inhabitants for the city to be considered
N_run = 5 # number of random catalog generated for computing the 2PCF
size = 7000 #number of points each random catalog has
k =  3  #number of time the 2PCF computing process is repeated to estimate variance
rmax = 1.6e5 #maximal distance between two points considered, if None, taken the max from data
scale = "log" #scale at which to bin the data, recommended "log"
nbins = 20 # number of bins to consider in the computation of 2PCF (betweem rmin and rmax)
bootstrap_size = 10_000
bootstrap_N_real = 10

parameters = {"threshold":threshold,"N_run":N_run,"size":size,"k":k,"rmax":rmax,"scale":scale,"nbins":nbins,"bootstrap_size":bootstrap_size,"bootstrap_N_real":bootstrap_N_real }

Nr_prime = size
N_sample = 2000

def plot_metadata(name, data_type, path_points_csv, path_border_geojson, Nsampl = N_sample,param = parameters, log_ax= False):
    def power_law(r, A, gamma):
        return A * r**(-gamma)

    gdf_points = mtpc.load_df_to_gdf(path_points_csv,threshold,  "population_density")

    if len(gdf_points) < 6*Nsampl:
        raise ValueError("Not enough point to split into 6 groups of size " + str(Nsampl))

    r, xi6, labels, l_av_bin_metadata = mtpc.compute_two_point_correlation_density_groups(
        gdf_points,
        path_border_geojson,
        name,
        param,
        Nr_prime,
        False,
        rmax,
        Nsampl
    )

    colors = ["blue", "blue", "blue", "orange", "orange", "orange"]
    linestyles = ["-", "--", ":", ":", "--", "-"]
        
    plt.figure(figsize=(10,6))

    for i in range(6):

        r_i = r
        xi_i = xi6[i][:-1]

        mask = (xi_i > 0) & (r_i > 0)
        r_fit = r_i[mask]
        xi_fit = xi_i[mask]

        # --- FIT LOG-LOG ---
        log_r = np.log(r_fit)
        log_xi = np.log(xi_fit)

        # régression linéaire : log(xi) = a + b log(r)
        b, a = np.polyfit(log_r, log_xi, 1)
        gamma = -b              # parce que b = -gamma
        A = np.exp(a)

        gamma_str = f"{gamma:.2f}"

        # --- PLOT ---
        plt.plot(
            r_i, 
            xi_i, 
            color=colors[i],
            linestyle=linestyles[i],
            label=f"{labels[i]} γ = {gamma_str}, pop_dens = {l_av_bin_metadata[i]:.2f}"
        )

    plt.xlabel("r")
    plt.ylabel("ξ(r)")
    plt.legend(
        loc='upper center',         # placer la légende au centre horizontalement
        bbox_to_anchor=(0.5, -0.15), # 0.5 = centre horizontal, -0.15 = en dessous du graphe
        ncol=2                      # nombre de colonnes dans la légende
    )
    plt.grid(True)
    if log_ax == True:
        plt.semilogy()
    plt.show()
```

```python
import numpy as np
import geopandas as gpd
from shapely.geometry import Point
from scipy.spatial.distance import cdist
import matplotlib.pyplot as plt
import matplotlib.cm as cm

def soneira_peebles_model_gdf_circle(lamb, eta, L, R, erase_nodes=False):
    rng = np.random.default_rng()
    position = [(0, 0)]  # point initial
    all_positions = []
    all_ages = []

    for i in range(L):
        new_positions = []
        new_ages = []
        for cx, cy in position:
            for _ in range(eta):
                # Génération uniforme dans un cercle
                theta = 2 * np.pi * rng.random()
                radius = (R / (lamb ** i)) * np.sqrt(rng.random())
                x_new = cx + radius * np.cos(theta)
                y_new = cy + radius * np.sin(theta)
                new_positions.append((x_new, y_new))
                new_ages.append(i)
        position = new_positions  # nouvelle génération
        all_positions += new_positions
        all_ages += new_ages

    # Effacement aléatoire optionnel
    if isinstance(erase_nodes, int) and erase_nodes > 0:
        temp_positions = list(all_positions)
        temp_ages = list(all_ages)
        while erase_nodes != 0 and len(temp_positions) > 0:
            index = rng.choice(range(len(temp_positions)))
            temp_positions.pop(index)
            temp_ages.pop(index)
            erase_nodes -= 1
        all_positions = temp_positions
        all_ages = temp_ages

    # Conversion en array pour calcul distance
    position_array = np.array(all_positions)
    distance_matrix = cdist(position_array, position_array)

    # GeoDataFrame des points
    gdf_points = gpd.GeoDataFrame({
        "age": all_ages,
        "geometry": [Point(x, y) for x, y in position_array]
    })

    # GeoDataFrame des cercles
    gdf_circles = gpd.GeoDataFrame({
        "age": all_ages,
        "geometry": [Point(x, y).buffer(R / (lamb ** age)) for (x, y), age in zip(all_positions, all_ages)]
    })

    return gdf_points, gdf_circles, distance_matrix


# =======================
# Exemple d'utilisation
# =======================
lamb, eta, L, R = 2, 2, 4, 1
gdf_points, gdf_circles, dist_matrix = soneira_peebles_model_gdf_circle(lamb, eta, L, R)

# Visualisation
fig, ax = plt.subplots(figsize=(8, 8))

ages = gdf_points["age"]
norm = plt.Normalize(vmin=ages.min(), vmax=ages.max())
cmap = cm.viridis

# Cercles
gdf_circles.plot(ax=ax, facecolor='none', edgecolor=cmap(norm(gdf_circles["age"])), linewidth=0.8)

# Points
sc = ax.scatter(
    [p.x for p in gdf_points.geometry],
    [p.y for p in gdf_points.geometry],
    c=gdf_points["age"],
    cmap=cmap,
    alpha=0.7
)

plt.colorbar(sc, label="Âge du nœud")
ax.set_aspect('equal')
plt.show()
```
