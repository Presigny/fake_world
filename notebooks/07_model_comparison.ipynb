## 0. Context

The 2PCF of many real spatial systems follows a power law ξ(r) = (r/r0)^(−γ). The
Soneira-Peebles model reproduces it by construction, with few parameters and a cost that
grows slowly with N. LGCP, Hawkes and Cox-Hawkes processes can also reproduce it, but
only after a costly inference. The SP mixture is therefore proposed as a realistic and
cheap initial placement of the nodes for spatial network models.

## 2. SP against the point processes at constant N

For each dataset we take **one subsample of N data points** (N capped at 10 000 and at
the size of the dataset), and each model generates N nodes inside the same border. We
compare the 2PCFs and the Jensen-Shannon divergence between them (**lower JSD means a
better match**).

Since the point processes are inferred by SVI, the comparison is fair, and the SP is not
expected to win everywhere. The aim is not to be the best 2PCF generator (Hawkes, LGCP
and Cox-Hawkes can also reproduce it) but to be competitive while having fewer
parameters and a much lower cost (section 3).

```python
def plot_2pcf_grid(cache):
    n = len(cache); ncol = 3; nrow = int(np.ceil(n / ncol))
    fig, axes = plt.subplots(nrow, ncol, figsize=(5 * ncol, 4 * nrow))
    axes = np.atleast_1d(axes).ravel()
    for ax, (name, c) in zip(axes, cache.items()):
        ax.plot(c["data_r"], c["data_xi"], "-o", color="k", lw=2, ms=3,
                label="data", zorder=5)
        for m in MODELS:
            if f"{m}_r" not in c or c[f"{m}_r"] is None:
                continue
            ax.plot(c[f"{m}_r"], c[f"{m}_xi"], "--", color=COLORS[m], alpha=0.8,
                    label=f"{m} (JSD={float(c[f'{m}_jsd']):.3f})")
        ax.set_xscale("log"); ax.set_yscale("log")
        ax.set_title(f"{name} (N={int(c['N'])})", fontsize=10)
        ax.set_xlabel("r (normalised)"); ax.set_ylabel(r"$\xi(r)$")
        ax.legend(fontsize=6.5); ax.grid(alpha=0.3)
    for ax in axes[len(cache):]:
        ax.axis("off")
    fig.tight_layout(); plt.show()


plot_2pcf_grid(CACHE)
```

## 2bis. Control on a neutral metric, nearest-neighbour distances

The JSD on the 2PCF is the home ground of the SP, since the SP is calibrated directly on
the 2PCF (slope γ and amplitude ρ), whereas the BSTPP models are fitted on the
**likelihood of the positions**, not on the 2PCF. For a less biased test we compare the
**distribution of nearest-neighbour (NN) distances**, a spatial summary that was not used
to calibrate the SP.

If the SP stays the best here, the result holds. If it drops, its advantage on the 2PCF
is an effect of the calibration.

```python
jsd_2pcf = pd.DataFrame({m: {n: float(c[f"{m}_jsd"]) for n, c in CACHE.items()}
                         for m in MODELS})
jsd_nn = pd.DataFrame({m: {n: float(c[f"nn_jsd_{m}"]) for n, c in CACHE.items()
                          if f"nn_jsd_{m}" in c} for m in MODELS})

print("=== mean rank (1 = best) ===")
comp = pd.DataFrame({
    "rank_2PCF (SP home ground)": jsd_2pcf.rank(axis=1).mean(),
    "rank_NN (neutral)": jsd_nn.rank(axis=1).mean(),
}).sort_values("rank_NN (neutral)")
display(comp.round(2))

print("\nNN JSD per dataset:")
display(jsd_nn.round(4))
jsd_nn.to_csv(ROOT / "out" / "07_jsd_nn.csv")

sp_nn_rank = jsd_nn.rank(axis=1)["SP"]
print(f"\nSP on the NN metric: best in {(sp_nn_rank == 1).sum()}/{len(sp_nn_rank)} datasets "
      f"(against {(jsd_2pcf.rank(axis=1)['SP'] == 1).sum()}/{len(jsd_2pcf)} on the 2PCF).")
```

```python
fig, ax = plt.subplots(figsize=(9, 4.5))
x = np.arange(len(MODELS))
r2 = jsd_2pcf.rank(axis=1).mean()[MODELS]
rn = jsd_nn.rank(axis=1).mean()[MODELS]
ax.bar(x - 0.2, r2, 0.4, label="2PCF (SP home ground)", color="#d62728", alpha=0.85)
ax.bar(x + 0.2, rn, 0.4, label="NN (neutral)", color="#1f77b4", alpha=0.85)
ax.set_xticks(x); ax.set_xticklabels(MODELS, rotation=15)
ax.set_ylabel("mean JSD rank (1 = best)")
ax.set_title("Mean rank of each model on the 2PCF and on the NN distances")
ax.legend(); ax.grid(alpha=0.3, axis="y")
plt.tight_layout(); plt.show()

have = [(n, c) for n, c in CACHE.items() if "nn_r" in c]
ncol = 3; nrow = int(np.ceil(len(have) / ncol))
fig, axes = plt.subplots(nrow, ncol, figsize=(5 * ncol, 3.6 * nrow))
axes = np.atleast_1d(axes).ravel()
for ax, (name, c) in zip(axes, have):
    ax.plot(c["nn_r"], c["nn_data"], "-o", color="k", ms=3, lw=2, label="data")
    for m in MODELS:
        if f"nn_{m}" in c:
            ax.plot(c["nn_r"], c[f"nn_{m}"], "--", color=COLORS[m], alpha=0.8, label=m)
    ax.set_xscale("log"); ax.set_xlabel("NN distance (norm.)"); ax.set_ylabel("density")
    ax.set_title(name, fontsize=10); ax.legend(fontsize=6.5); ax.grid(alpha=0.3)
for ax in axes[len(have):]:
    ax.axis("off")
fig.suptitle("Distribution of the nearest-neighbour distances", y=1.0)
plt.tight_layout(); plt.show()
```

### Why does the SP match the NN distances on some systems only?

The SP wins on the NN metric for the large structured networks, but not for the small
city datasets. The hypothesis tested here is that **the NN fidelity of the SP depends on
the size of the system**. The SP is a hierarchical fractal. On a **small** system it has
few levels, so it over-clusters (too many very close pairs), which gives an excess of
short NN distances and a poor JSD. On a **large** system it has enough levels to spread
the multi-scale structure, and the NN distribution matches better.

- **Left panel.** NN JSD of the SP against the size N (log-log). A decreasing trend means
  that the SP reproduces the NN distances better on larger systems.
- **Right panel.** The mechanism. On a small system (Belgium) the SP over-produces short
  distances, on a large one (Germany) the gap disappears.

```python
from scipy.stats import spearmanr

rows = []
for name, c in CACHE.items():
    if "nn_r" not in c:
        continue
    N_sys = int(c["N_full"]) if "N_full" in c else int(c["N"])
    jsds = {m: float(c[f"nn_jsd_{m}"]) for m in MODELS if f"nn_jsd_{m}" in c}
    rows.append((name, N_sys, float(c["nn_jsd_SP"]), min(jsds, key=jsds.get) == "SP"))
diag = pd.DataFrame(rows, columns=["dataset", "N", "nn_jsd_SP", "sp_win"]
                    ).set_index("dataset")
rho_s = spearmanr(np.log(diag["N"]), diag["nn_jsd_SP"]).correlation

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
ax = axes[0]
for win, sub in diag.groupby("sp_win"):
    ax.scatter(sub["N"], sub["nn_jsd_SP"], s=80, zorder=3,
               c=("#2ca02c" if win else "#d62728"),
               label="SP best on NN" if win else "SP beaten")
for name, row in diag.iterrows():
    ax.annotate(name.replace(" Cities", "").replace(" France", "").replace(" Stops", ""),
                (row["N"], row["nn_jsd_SP"]), fontsize=7, ha="center", va="bottom")
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlabel("system size N (log)"); ax.set_ylabel("NN JSD of the SP (log)")
ax.set_title(f"NN JSD of the SP against the system size\n"
             f"Spearman(log N, NN JSD) = {rho_s:.2f}")
ax.legend(); ax.grid(alpha=0.3, which="both")

ax = axes[1]
for name, sty, lab in [("Belgium Cities", "-", "small (Belgium)"),
                       ("Germany Stops", "--", "large (Germany)")]:
    if name not in CACHE:
        continue
    c = CACHE[name]
    ax.plot(c["nn_r"], c["nn_data"], sty, color="k", lw=2, label=f"{lab}, data")
    ax.plot(c["nn_r"], c["nn_SP"], sty, color="#d62728", lw=1.6, label=f"{lab}, SP")
ax.set_xscale("log"); ax.set_xlabel("NN distance (norm.)"); ax.set_ylabel("density")
ax.set_title("On the small system the SP over-produces\nshort distances, not on the large one")
ax.legend(fontsize=8); ax.grid(alpha=0.3)
plt.tight_layout(); plt.show()

print(f"Spearman correlation (log N, NN JSD of the SP) = {rho_s:.2f}")
```

## 4. Comparison at constant N, sweep over N

`compare_at_constant_N` takes **one subsample of N data points**, computes its 2PCF, and
each model generates **N nodes** inside the same border, the JSD being measured against
this reference. Data and models are therefore all at the same N, which separates the
effect of the model from the effect of the sample size.

Here N is swept over {200, 500, 1000, 2000, 4000} (capped at N_real) on the datasets that
are not too large, to check that the JSD ranking is stable when N changes. The three
largest datasets (ENEDIS, France Routes, Germany) are left out of the sweep, since
reloading them takes about 30 min. They stay in the comparison of section 2 at N = 3000.

```python
def plot_sweep_grid(cache):
    have = [(n, c) for n, c in cache.items() if "sweep_N" in c]
    if not have:
        print("no sweep in the cache (run scripts/compute_all.py)."); return
    ncol = 3; nrow = int(np.ceil(len(have) / ncol))
    fig, axes = plt.subplots(nrow, ncol, figsize=(5 * ncol, 4 * nrow))
    axes = np.atleast_1d(axes).ravel()
    for ax, (name, c) in zip(axes, have):
        Ns = c["sweep_N"]
        for m in MODELS:
            if f"sweep_jsd_{m}" not in c:
                continue
            ax.plot(Ns, c[f"sweep_jsd_{m}"], "o-", color=COLORS[m], label=m)
        ax.set_xscale("log"); ax.set_yscale("log")
        ax.set_xlabel("N"); ax.set_ylabel("JSD")
        ax.set_title(name, fontsize=10); ax.grid(alpha=0.3, which="both")
        ax.legend(fontsize=6.5)
    for ax in axes[len(have):]:
        ax.axis("off")
    fig.suptitle("JSD against N", y=1.0)
    fig.tight_layout(); plt.show()


plot_sweep_grid(CACHE)

wins = tot = 0
for name, c in CACHE.items():
    if "sweep_N" not in c:
        continue
    for i in range(len(c["sweep_N"])):
        jsds = {m: c[f"sweep_jsd_{m}"][i] for m in MODELS
                if f"sweep_jsd_{m}" in c and np.isfinite(c[f"sweep_jsd_{m}"][i])}
        if jsds and min(jsds, key=jsds.get) == "SP":
            wins += 1
        tot += 1
if tot:
    print(f"SP is the best model in {wins}/{tot} points (dataset × N) of the sweep.")
```

## 5. Stochasticity and reproducibility (10 realisations per model)

Every model is random. We generate **10 independent realisations** of each model and
measure

- the mean **2PCF band** ± σ (how much the shape varies from one draw to the next),
- the **mean JSD ± σ**, which says how well a model matches on average and also how
  reproducible it is. A model with a low σ_JSD is reliable, which is a **second
  selection criterion** next to the mean JSD.

```python
def plot_sp_band_grid(cache):
    have = [(n, c) for n, c in cache.items() if c.get("stoch_r") is not None]
    ncol = 3; nrow = int(np.ceil(len(have) / ncol))
    fig, axes = plt.subplots(nrow, ncol, figsize=(5 * ncol, 4 * nrow))
    axes = np.atleast_1d(axes).ravel()
    for ax, (name, c) in zip(axes, have):
        r, mu, sd = c["stoch_r"], c["stoch_xi_mean"], c["stoch_xi_std"]
        ax.plot(c["data_r"], c["data_xi"], "-o", color="k", lw=2, ms=3,
                label="data", zorder=5)
        ax.plot(r, mu, "-", color=COLORS["SP"], lw=2, label="SP mean")
        ax.fill_between(r, np.clip(mu - sd, 1e-4, None), mu + sd,
                        color=COLORS["SP"], alpha=0.25, label=r"$\pm\sigma$")
        ax.set_xscale("log"); ax.set_yscale("log")
        ax.set_title(name, fontsize=10)
        ax.set_xlabel("r"); ax.set_ylabel(r"$\xi(r)$")
        ax.legend(fontsize=7); ax.grid(alpha=0.3)
    for ax in axes[len(have):]:
        ax.axis("off")
    fig.tight_layout(); plt.show()


plot_sp_band_grid(CACHE)
```

```python
jm = pd.DataFrame({m: {n: float(c[f"band_{m}_jsd_mean"]) for n, c in CACHE.items()
                       if f"band_{m}_jsd_mean" in c} for m in MODELS})
js = pd.DataFrame({m: {n: float(c[f"band_{m}_jsd_std"]) for n, c in CACHE.items()
                       if f"band_{m}_jsd_std" in c} for m in MODELS})

repro = pd.DataFrame({
    "mean JSD (accuracy)": jm.mean(),
    "sigma_JSD (reproducibility)": js.mean(),
}).loc[MODELS]
print("average over the datasets, accuracy and reproducibility:")
display(repro.round(4))
repro.to_csv(ROOT / "out" / "07_reproducibility.csv")

fig, ax = plt.subplots(figsize=(7, 5))
for m in MODELS:
    ax.scatter(jm[m].mean(), js[m].mean(), s=90, color=COLORS[m], zorder=3)
    ax.annotate(m, (jm[m].mean(), js[m].mean()), fontsize=9,
                xytext=(5, 5), textcoords="offset points")
ax.set_xlabel("mean JSD  (← more accurate)")
ax.set_ylabel(r"$\sigma_{JSD}$  (← more reproducible)")
ax.set_title("Accuracy against reproducibility (bottom left is best)")
ax.grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

## 6. Population marks, hierarchical SP against a marked Hawkes process

Real nodes are not interchangeable. Each one carries a **mark** (population of a city,
population density around a substation, ...). A good generative model should reproduce
where the nodes are (the 2PCF) and also how the mark is distributed in space. Typically
the large nodes (big cities, hubs) are rare, central, and spatially correlated with each
other. We therefore (a) assign a population to the generated nodes and (b) check that it
**correlates with the spatial structure** as in the data.

**How the SP handles it, the hierarchy.** The SP is a **tree**. The root has η children,
each of them η grandchildren, and so on. Keeping the *full* tree (and not only the leaves
used for the 2PCF) gives a natural hierarchy, with few nodes at the first levels (central,
structuring, the large cities) and many at the deep levels (peripheral, small units). We
assign

$$\text{population}(\text{node}) \;\propto\; \eta^{-\,\text{slope}\cdot\text{level}}$$

which gives by construction a power-law rank-size distribution (Zipf-like). The `slope`
is **calibrated** so that the rank-size exponent of the SP matches the one of the data
(`fit_zipf_exponent` and `calibrate_sp_slope`).

**The competitor, a marked Hawkes process.** To avoid judging the SP alone, we add a
point-process baseline, a **marked Hawkes process** (`simulate_marked_hawkes`) in which
the population of each node **modulates its self-excitation**, so that a large node has
more children nearby. The marks of the background points are drawn from the empirical
population distribution.

**Two validation measures** (data, marked SP, marked Hawkes). We test whether a large
population means a more clustered position.

1. **Binary split** (γ_hi against γ_lo). High (top 20 %) against low population, and
   comparison of the slope γ. A coupling gives **γ_hi > γ_lo**.
2. **Continuous mark correlation** $k_{mm}(r)$ (Stoyan). Each pair is weighted by the
   product of the populations, $k_{mm}(r)=\langle m_i m_j\rangle_{r}/\langle m\rangle^2$.
   $k_{mm}(r)>1$ means that large populations are close to each other, $k_{mm}=1$ means
   no coupling.

```python
zipf = pd.DataFrame([
    dict(dataset=n, zipf_data=round(float(c["zipf_data"]), 2),
         zipf_SP=round(float(c["zipf_sp"]), 2), slope_SP=round(float(c["sp_slope"]), 2))
    for n, c in CACHE.items() if "zipf_data" in c]).set_index("dataset")
print("rank-size (Zipf) exponent, the SP slope is fitted on the data:")
display(zipf)

mk_rows = []
for name, c in CACHE.items():
    if "mh_gamma_hi" not in c:
        continue
    row = {"dataset": name}
    for tag in ("data", "sp", "mh"):
        row[f"{tag}_hi"] = float(c[f"{tag}_gamma_hi"])
        row[f"{tag}_lo"] = float(c[f"{tag}_gamma_lo"])
    mk_rows.append(row)
mk = pd.DataFrame(mk_rows).set_index("dataset")

fig, axes = plt.subplots(1, 3, figsize=(16, 4.5), sharey=True)
for ax, (lab, tag) in zip(axes, [("data", "data"), ("marked SP", "sp"),
                                 ("marked Hawkes", "mh")]):
    x = np.arange(len(mk))
    ax.bar(x - 0.2, mk[f"{tag}_hi"], 0.4, label="high population", color="#c0392b")
    ax.bar(x + 0.2, mk[f"{tag}_lo"], 0.4, label="low population", color="#3498db")
    ax.set_xticks(x); ax.set_xticklabels(mk.index, rotation=40, ha="right", fontsize=7)
    ax.set_title(lab); ax.set_ylabel(r"$\gamma$"); ax.legend(fontsize=8)
    ax.grid(alpha=0.3, axis="y")
fig.suptitle("Clustering of high and low population nodes (coupling means γ_hi > γ_lo)")
plt.tight_layout(); plt.show()

dgamma = pd.DataFrame({
    "data": mk["data_hi"] - mk["data_lo"],
    "marked SP": mk["sp_hi"] - mk["sp_lo"],
    "marked Hawkes": mk["mh_hi"] - mk["mh_lo"],
})
print("\nΔγ = γ_hi − γ_lo (population-structure coupling, > 0 expected):")
display(dgamma.round(2))
print("correlation of the coupling, model against data (Δγ):")
print(f"  marked SP     : {dgamma['data'].corr(dgamma['marked SP']):.2f}")
print(f"  marked Hawkes : {dgamma['data'].corr(dgamma['marked Hawkes']):.2f}")
```

```python
def plot_markcorr_grid(cache):
    have = [(n, c) for n, c in cache.items() if "markcorr_r" in c]
    if not have:
        print("no mark correlation in the cache for these datasets."); return
    ncol = 3; nrow = int(np.ceil(len(have) / ncol))
    fig, axes = plt.subplots(nrow, ncol, figsize=(5 * ncol, 4 * nrow))
    axes = np.atleast_1d(axes).ravel()
    for ax, (name, c) in zip(axes, have):
        r = c["markcorr_r"]
        ax.plot(r, c["data_kmm"], "-o", color="k", ms=3, label="data")
        ax.plot(r, c["sp_kmm"], "--s", color=COLORS["SP"], ms=3, label="marked SP")
        if "mh_kmm" in c:
            ax.plot(r, c["mh_kmm"], "--^", color=COLORS["Hawkes"], ms=3,
                    label="marked Hawkes")
        ax.axhline(1.0, color="gray", ls=":", lw=1)     # k_mm = 1, no coupling
        ax.set_xscale("log")
        ax.set_xlabel("r (normalised)"); ax.set_ylabel(r"$k_{mm}(r)$")
        ax.set_title(name, fontsize=10); ax.legend(fontsize=7); ax.grid(alpha=0.3)
    for ax in axes[len(have):]:
        ax.axis("off")
    fig.suptitle(r"mark correlation $k_{mm}(r)$  "
                 r"($>1$: large populations close to each other)", y=1.0)
    fig.tight_layout(); plt.show()


plot_markcorr_grid(CACHE)
```

## 7. Summary

| Question | Section | Result |
|----------|---------|--------|
| SP against point processes at constant N (inferred by SVI) | 2 | SP competitive, not dominant |
| Control on a neutral metric (NN distances) | 2bis | SP best mostly on the large systems |
| Computational cost | 3 | SP inference faster by several orders of magnitude |
| Comparison at constant N, sweep over N | 4 | `sweep_constant_N` |
| Stochasticity and reproducibility | 5 | `stochastic_bands`, JSD ± σ |
| Population marks, hierarchical SP against marked Hawkes | 6 | calibrated Zipf slope, $k_{mm}$ |

Since the inferred point processes close most of the JSD gap, the arguments for the SP
are its **low cost** (section 3), its **reproducibility** (section 5), and a
**population coupling that comes from the hierarchy itself** (section 6), where a Hawkes
process needs the mark to be added by hand.
